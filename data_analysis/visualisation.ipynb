```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
mapd_data = {}
mapd_data["ws100"] = pd.read_csv('ws100_data.csv')
mapd_data["ws200"] = pd.read_csv('ws200_data.csv')
mapd_data["ws300"] = pd.read_csv('ws300_data.csv')
mapd_data["ws400"] = pd.read_csv('ws400_data.csv')
mapd_data["ws500"] = pd.read_csv('ws500_data.csv')
mapd_data["ws600"] = pd.read_csv('ws600_data.csv')
```

```python
for mapd_key in mapd_data.keys():
  df_mapd = mapd_data[mapd_key]

  # Scatter Plot
  df_melted = df_mapd.melt()
  x = df_melted.variable.tolist()
  y = df_melted.value.tolist()
  plt.scatter(x, y, c='red')

  # Trend Line
  columns = df_mapd.columns.tolist()
  means = df_mapd.mean()
  plt.plot(columns,means,c='black',alpha=0.7,linestyle=':')

  # Design
  plt.grid(True,axis='y',linestyle='--', alpha=0.5)
  plt.xlabel("Number of Network Timesteps")
  plt.ylabel("Throughput (t=1000)")
  title = str(mapd_key).upper()
  plt.title(f"Throughput of {title} using Time-Expanded Network Flows")

  plt.show()
```
